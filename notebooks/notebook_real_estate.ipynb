## 🔍 OVERVIEW

```python
def inspect_df(df, name=None, n=5):
    title = f"DataFrame Inspection {name}" if name else "DataFrame Inspection"

    display(Markdown(f"l🔍 {title}"))
    display(Markdown("----"))

    display(Markdown("## 📏 Dimensions"))
    display(df.shape)

    display(Markdown("## ℹ️ DF Info"))
    display(df.info())

    display(Markdown(f"## 👀 {n} First Rows"))
    display(df.head(n))

    display(Markdown(f"## 👀 {n} Last Rows"))
    display(df.tail(n))

    display(Markdown("## 🎲 {n} Ramdom Rows"))
    display(df.sample(n), random_state=42)

    display(Markdown("## 📊 Descriptive Stats"))
    display(df.describe(include="all").T)

    display(Markdown("## 🧬 Data Type"))
    display(df.dtypes.to_frame("dtypes"))

    display(Markdown("## 🔢 Unique Values"))
    display(df.nunique().sort_values(ascending=False).to_frame("n_unique"))

    display(Markdown("## ⚠️ NaN Values"))
    display(df.isna().sum().sort_values(ascending=False).to_frame("n_missing"))

    display(Markdown("## ⚠️ NaN Values (%)"))
    display((df.isna().mean() * 100).round(2).sort_values(ascending=False).to_frame("pct_missing"))

    display(Markdown("## 🔂 Duplicate Rows Count"))
    display(df.duplicated().sum())

    display(Markdown("## 🔁 Duplicate Row"))
    display(df[df.duplicated()])
```

```python
inspect_df(df, name="df", n=5)
display(Markdown("----"))
```

## PREPROCESSING

### Feature Engineering

```python
from sklearn.pipeline import make_pipeline
from sklearn.impute import SimpleImputer
from sklearn.preprocessing import RobustScaler
from category_encoders import TargetEncoder
```

```python
df = df[df["status"] == "sold"].copy()
df["prev_sold_year"] = (df["prev_sold_date"].dt.year.astype("float64"))
df = df.dropna(subset=["price"]).copy() # Delete all price missing value

X = df.drop(columns=["brokered_by", "zip_code","acre_lot", "street", "log_price", "prev_sold_date", "price", "status"])
y = df["price"]
```

```python
display (X.isna().sum().sort_values(ascending=False).to_frame("X_missing"), len(X), X.dtypes.to_frame("X_types"), y.isna().sum(),)
```

#### Categoricals features

```python
# most_frequent → classic & simple

cat_transformer = make_pipeline(
    SimpleImputer(strategy="most_frequent"),
    TargetEncoder(), # Instead of creating 1 column per city, it'll replace each city name with the mean price of properties in that city
)

cat_cols = X.select_dtypes(include=["object", "str"]).columns.tolist()
```

#### Numericals features

```python
num_transformer = make_pipeline(
    SimpleImputer(strategy="median"),
    RobustScaler()
)

num_cols = X.select_dtypes(include=["float64", "int64", "Int64"]).columns.tolist()
```

#### Setup Baseline

```python
from sklearn.compose import make_column_transformer
from sklearn.linear_model import LinearRegression
```

```python
preprocessing = make_column_transformer(
    (cat_transformer, cat_cols),
    (num_transformer, num_cols),
).set_output(transform="default") # sparse_out=True
```

```python
model = LinearRegression()
```

```python
pipe_baseline = make_pipeline(
    (preprocessing),
    model,
)
```

## TRAINING

### Train Test Split

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test, y_train, y_test = train_test_split(
    X,
    y,
    test_size=0.3,
    random_state=42,
)

baseline = pipe_baseline.fit(X_train, y_train)

# X_train_sample = X_train.sample(
#     n=min(100_000, len(X_train)),
#     random_state=42,
# )
# y_train_sample = y_train.loc[X_train_sample.index]

# pipe_baseline.fit(X_train_sample, y_train_sample)
```

### Cross Validate

```python
from sklearn.metrics import get_scorer_names
from sklearn.model_selection import cross_validate, KFold
```

```python
get_scorer_names()
```

```python
scoring = {
        "r2":"r2",
        "mae": "neg_mean_absolute_error",
        "mse": "neg_mean_squared_error",
        "rmse": "neg_root_mean_squared_error",
        "max_error": "neg_max_error",
        }

cv_splits = KFold(
    n_splits=5,
    shuffle=True,
    random_state=42,

)
scores_baseline = cross_validate(
    pipe_baseline,
    X_train,
    y_train,
    cv=cv_splits,
    scoring=scoring,
    n_jobs=-1,

)
```

```python
cv_scores = pd.DataFrame(scores_baseline)
cv_scores
```

```python
results = {
    "R2": np.mean(scores_baseline["test_r2"]),
    "MAE": -np.mean(scores_baseline["test_mae"]),
    "MSE": -np.mean(scores_baseline["test_mse"]),         # negated because cross-validate returns negative MSE
    "RMSE": -np.mean(scores_baseline["test_rmse"]),
    "MAX ERROR": -np.mean(scores_baseline["test_max_error"]),  # negated because scorer is negative
}

cv_results = pd.DataFrame([results])
cv_results
```

### Best Params

```python
from sklearn.linear_model import Ridge, Lasso
from sklearn.model_selection import RandomizedSearchCV
```

```python
param_grid = [
    {
        "linearregression": [LinearRegression()],
    },

    {
        "linearregression": [Ridge()],
        "linearregression__alpha": [0.01, 0.1, 1, 10, 100, 1000],
        "linearregression__solver": ['sag', 'lsqr', 'sparse_cg'],
        "linearregression__max_iter": [1_000, 2_000, 3_000, 4_000, 10_000],
    },

    {
        "linearregression": [Lasso()],
        "linearregression__alpha": [0.01, 0.1, 1, 10, 100, 1000],
        "linearregression__selection": ["cyclic", "random"],
        "linearregression__max_iter": [1_000, 2_000, 3_000, 4_000, 10_000],
    }
]

cv_splits_search = KFold(
    n_splits=3,
    shuffle=True,
    random_state=42,

)

scoring_search = {
    "r2":"r2",
    "mae": "neg_mean_absolute_error",
    "mse": "neg_mean_squared_error",
    "rmse": "neg_root_mean_squared_error",
    "max_error": "neg_max_error",
    }

search = RandomizedSearchCV(
    pipe_baseline,
    param_distributions=param_grid,
    cv=cv_splits_search,
    scoring=scoring_search,
    refit="rmse",
    n_jobs=-1,
    error_score="raise",
    verbose=1,
    random_state=42,
)

X_train_sample = X_train.sample(
    n=min(100_000, len(X_train)),
    random_state=42,
)

y_train_sample = y_train.loc[X_train_sample.index]

search.fit(X_train_sample, y_train_sample)
```

```python
display(search.best_estimator_, search.best_params_, -search.best_score_)
```

```python
search_results = pd.DataFrame(search.cv_results_)
search_results
```

```python
results = -search_results[
    ["mean_test_r2", "mean_test_rmse", "mean_test_mae", "mean_test_max_error"]
].sort_values("mean_test_rmse", ascending=False)

results = pd.DataFrame(results)
results
```

```python
best_model = search.best_estimator_
best_model.fit(X_train, y_train)
```

## CHECK RESULTS

## ORAL

## Question 1
### **Tu as utilisé RMSE comme métrique principale. Pourquoi pas MAE ?**

`RMSE pénalise davantage les grandes erreurs — c'est ce qu'on veut en immobilier où une grosse erreur de prix peut faire rater une vente ou survendre un bien. La MAE traiterait une grosse erreur de la même façon qu'une erreur petite erreur.`

## Question 2
### **Ton R² est à 0.25. C'est bon ou mauvais ? Et qu'est-ce que ça veut dire concrètement ?**

`La LinearRegression suppose une relation linéaire entre les features et le prix. Sur ce dataset les relations sont non-linéaires — un bien avec 5 salles de bain ne coûte pas forcément 5x plus cher qu'un bien avec 1 salle de bain. RandomForest et XGBoost captent ces relations non-linéaires et les interactions entre variables.`

## Question 3
### **Tu as filtré sur status == "sold" avant d'entraîner. Pourquoi ? Qu'est-ce qui se passerait si tu avais gardé les biens for_sale ?**

` Parce que `for_sale` est le prix demandé, et le vendeur peut demander n'importe quel prix. Un prix `sold` c'est un prix validé par le marché. Entrainer sur des prix `for_sale`introduirait du bruit et biaiserait le modèle vers des prix sous et surestimés. `

## Question 4
### **Tu as créé prev_sold_year depuis prev_sold_date. Pourquoi cette variable pourrait être utile pour prédire le prix ?**

`Un bien vendu en 2023 a un  prix qui reflète le marché actuel. L'année de vente capture indirectement la dynamique du marché de l'immobilier.`
